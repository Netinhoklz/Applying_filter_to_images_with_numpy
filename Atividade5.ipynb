## Tratando a imagem para a escala de cinza

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

imagem = cv2.imread('estatua_do_jose_walter.jpg')
imagem = cv2.cvtColor(imagem,cv2.COLOR_BGR2GRAY)

plt.imshow(imagem,cmap = 'gray')
plt.xticks([])
plt.yticks([])
plt.show()
```

# Questão 01
- Implemente seguintes filtros mostrados na aula:

## a. Filtro laplaciano

```python
filtro = np.array([
    [1, 1, 1],
    [1, -8, 1],
    [1, 1, 1]
])

altura_k, largura_k = filtro.shape
pad_h, pad_w = altura_k // 2, largura_k // 2
imagem_pad = np.pad(imagem, ((pad_h, pad_h), (pad_w, pad_w)), mode='constant', constant_values=0)
resultado_corr = np.zeros_like(imagem, dtype=int)

for y in range(imagem.shape[0]):
    for x in range(imagem.shape[1]):
        regiao = imagem_pad[y:y+altura_k, x:x+largura_k]
        resultado_corr[y, x] = np.sum(regiao * filtro)

resultado_corr = np.clip(resultado_corr, 0, 255).astype(np.uint8)

fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(16, 9))

ax1.set_title('Imagem normal')
ax1.imshow(imagem, cmap = 'gray')
ax1.axis('off')

ax2.set_title('Imagem com filtro aplicado')
ax2.imshow(resultado_corr, cmap='gray')
ax2.axis('off')

plt.show()
```

# Plotando as distribuição da imagem

```python
plt.figure()                       # cria uma nova figura
plt.hist(imagem.ravel(),           # vetoriza a imagem
         bins=256,                 # um “balde” por nível de cinza
         range=(0, 255))           # intervalo completo de intensidades
plt.title('Histograma da imagem em escala de cinza')
plt.xlabel('Intensidade de cinza (0–255)')
plt.ylabel('Número de pixels')
plt.tight_layout()
plt.show()
```

## Aplicando OTSU

```python
# Algoritmo de Otsu
hist = np.bincount(imagem.ravel(), minlength=256).astype(np.float64)
total = imagem.size
prob = hist / total

omega = np.cumsum(prob)                   # ω_k
mu = np.cumsum(prob * np.arange(256))     # μ_k
mu_t = mu[-1]                             # μ_T

epsilon = 1e-10
sigma_b2 = (mu_t * omega - mu) ** 2 / (omega * (1 - omega) + epsilon)
thresh = int(np.argmax(sigma_b2))

# Binariza
imagem_bin = (imagem >= thresh).astype(np.uint8) * 255
print(f"Valor de corte T: {thresh}")
resultado_corr = np.where(imagem_bin, 0, 255).astype(np.uint8)

fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(16, 9))

ax1.set_title('Imagem normal')
ax1.imshow(imagem, cmap = 'gray')
ax1.axis('off')

ax2.set_title('Imagem com filtro aplicado')
ax2.imshow(resultado_corr, cmap='gray')
ax2.axis('off')

plt.show()
```
